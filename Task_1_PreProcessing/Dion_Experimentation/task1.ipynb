<h2>Imports</h2>

```python
import tensorflow as tf 
import matplotlib.pyplot as plt
import cv2 
```

<p>preprocess image ensures that image passed in, no matter the size is reshaped and fit for model consumption, in proper paths are hamdled</p>

```python
def preprocess_image(path):
    image = cv2.imread(path)
    if image is None:
        raise ValueError("Error retrieving the image")
    else:
        image = cv2.resize(image, (28, 28))
        image = 255 - image
        image = tf.expand_dims(image, axis=0)
        plt.imshow(image[0], cmap='gray')
        return image
```
